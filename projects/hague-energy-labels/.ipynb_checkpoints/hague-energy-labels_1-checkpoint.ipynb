## Executive summary

| **−0.14** | **+0.47** | **71% vs 40%** | **51% vs 6%** |
|---|---|---|---|
| Rank correlation of income with the share of A-label homes, all neighbourhoods (n = 106, p = 0.16) · *Figures 9, 13* | Rank correlation of income with the share of E–G homes outside the new-build district (n = 91, p < 0.001) · *Figure 13* | Median share of E–G homes in high-income neighbourhoods vs the rest · *Figure 11* | Median share of A-label homes in the new-build district vs the rest of the city · *Figure 12* |

**Question.** Do neighbourhoods in The Hague with higher household incomes have more energy-efficient homes?

**Answer: no, and in the older city it is the other way round.** Across all 106 neighbourhoods with data, income has no clear link with the share of A-label homes. That near-zero result hides two opposite patterns:

* **Outside the new-build district**, richer neighbourhoods have clearly *worse* labels. The wealthy villa areas along the dunes (Duttendel, Marlot, Nassaubuurt, Vogelwijk) have 86–94% of homes at E, F or G.
* **The best labels are in Leidschenveen-Ypenburg**, built after 1995, where incomes are close to the city average. Labels follow the **age of the housing**, not the income of the people living in it.

The high-income quarter of neighbourhoods is worse on all three label measures (Mann-Whitney p = 0.002 to 0.022). Income and labels are both strongly clustered on the map (Moran's I ≈ 0.6), so wealthy old housing and new energy-efficient housing sit in separate parts of the city.

**Part II, neighbourhood types.** K-means on the label mix finds **five stable types**: Mostly A, Mostly A and B, Mixed (centred on C), E-dominated and G-dominated. The two poorest types cover half of all neighbourhoods. The types form clear zones on the map (50% of neighbouring pairs share a type, against 26% by chance), and the two A types have middle incomes, while 13 of the 27 high-income neighbourhoods are in the G-dominated type.

## 2. Load and profile the data

Two inputs:

| File | Contents |
|---|---|

| `neighborhoods.json` (or `.shp`) | Boundaries of the 114 neighbourhoods (*buurten*), prepared for the course, in WGS 84 |

The **standardised disposable income** is household income after tax and transfers, corrected for household size and composition, so a single person and a family of four can be compared.

> **Why this step.** Before any analysis it is worth checking three things: that the columns are what their names say (the CSV lists label G *before* F), that every row is a real neighbourhood, and that every neighbourhood in the table finds its shape on the map. A silent mismatch in the join would drop areas from the maps without any error.

````python
CSV_FILE = find_file("neighborhoods.csv", "*neighborhoods*.csv")

raw = pd.read_csv(CSV_FILE, index_col=0)

raw.index.name = "neighbourhood"

file_order = [c.replace("Energielabel ", "") for c in raw.columns[:-1]]

print(f"{CSV_FILE.name}: {raw.shape[0]} rows × {raw.shape[1]} columns")

print("Label columns in file order:", file_order, "+ income")

raw.tail(3)
````

**What is in the file?** Homes per label for the whole table, and the income value of every row.

````python
label_col = {c.replace("Energielabel ", ""): c for c in raw.columns[:-1]}

homes_per_label = pd.Series({lab: raw[label_col[lab]].sum() for lab in LABELS})

income_raw = raw[raw.columns[-1]]

unknown_income = income_raw["Unknown"]


fig, (ax, ax2) = plt.subplots(1, 2, figsize=(14, 4.4), width_ratios=[1.1, 1])

# left: homes per label, columns picked by name (A++ ... G), not by position
ax.bar(LABELS, homes_per_label / 1000, color=LABEL_COLORS, width=0.75)

for i, v in enumerate(homes_per_label):
    ax.text(i, v / 1000 + 1, f"{v / 1000:,.0f}k" if v >= 1000 else f"{v:,.0f}", ha="center", fontsize=8, color=MUTED)

ax.set_ylabel("Homes with this label (thousands)")

ax.set_title(f"{homes_per_label.sum():,.0f} homes with a registered label")

ax.text(0, -0.2, "File column order: " + " · ".join(file_order) + "   →   reordered by name",
        transform=ax.transAxes, fontsize=8, color=MUTED)

ax.grid(axis="x", visible=False)

# right: income of every row, with the 'Unknown' row marked
bins = np.arange(-6000, income_raw.max() + 4000, 2500)

ax2.hist(income_raw.drop("Unknown").dropna(), bins=bins, color="#b9b8b2")

ax2.axvline(unknown_income, color=ORANGE, lw=2.5)

ax2.text(unknown_income + 1200, ax2.get_ylim()[1] * 0.85,
         f"'Unknown' row: −€{abs(unknown_income):,.0f}\n(a balancing figure, not a neighbourhood)",
         color=ORANGE, fontsize=8.5)

ax2.xaxis.set_major_formatter(eur)

ax2.set_xlabel("Average standardised disposable household income")

ax2.set_ylabel("Rows")

ax2.set_title(f"Income per row · {income_raw.drop('Unknown').isna().sum()} neighbourhoods have no figure")

save_figure(fig, "data_profile", "The table covers 261,797 labelled homes; one row is not a neighbourhood", y=1.03, tight=True)
````

> **Caution.** The last row, `Unknown`, is not a neighbourhood: it holds homes that could not be placed in a *buurt*, and its income value (−3,407) is a balancing figure, not a real average (Figure 1, right). It is dropped below. Note also that the export puts **G before F**; the code reorders the columns by name, never by position (Figure 1, left).

> **Insight.** All 114 neighbourhoods in the table match a shape on the map, with no leftovers on either side (Figure 2). ## 3. Clean the data and define the measures

**Measures used throughout**

| Measure | Definition |
|---|---|
| **% A-label or better** | Homes with label A++, A+ or A, as a share of all homes with a registered label |
| **% E–G** | Homes with label E, F or G, as a share of all labelled homes |
| **Average label score** | Mean of the labels on a 1–9 scale (G = 1, F = 2 … A = 7, A+ = 8, A++ = 9). One number that uses all nine classes, not just the two ends |
| **High income** | Above the 75th percentile of neighbourhood income in The Hague, as stated in the original hypothesis |
| **New-build district** | Leidschenveen-Ypenburg (neighbourhood codes 101–121): added to The Hague in 2002 and built mostly from the late 1990s onward |

**Hypothesis (H1).** Neighbourhoods with higher household income have a larger share of homes with good energy labels.

> **Why this step.** The original notebook grouped labels into three bands (high, medium, low). That is kept, but a single **average label score** is added, because a neighbourhood that moves from G to D is better off even though it never reaches the high band. The **new-build** flag is added because the age of the housing is the most likely confounder: new homes get good labels whoever lives in them.

````python
df = raw.drop(index="Unknown").copy()

df.columns = [c.replace("Energielabel ", "") for c in df.columns[:-1]] + ["income"]

df = df[LABELS + ["income"]]                      # reorder by name: A++ … G


# label shares and summary measures
df["labelled_homes"] = df[LABELS].sum(axis=1)

shares = df[LABELS].div(df["labelled_homes"], axis=0) * 100

df["pct_A_or_better"] = shares[["A++", "A+", "A"]].sum(axis=1)

df["pct_B_to_D"] = shares[["B", "C", "D"]].sum(axis=1)

df["pct_E_to_G"] = shares[["E", "F", "G"]].sum(axis=1)

df["label_score"] = (df[LABELS] * pd.Series(LABEL_SCORE)).sum(axis=1) / df["labelled_homes"]


# neighbourhood code and the new-build flag
df["code"] = df.index.str.split(" ").str[0].astype(int)

df["area"] = np.where(df["code"] >= 101, NEW_BUILD, REST)


print(f"{len(df)} neighbourhoods · {df['labelled_homes'].sum():,.0f} homes with a registered label")

print(f"Income missing for {df['income'].isna().sum()} neighbourhoods")

df.sort_values("labelled_homes")[["labelled_homes", "income", "pct_A_or_better"]].head(8)
````

> **Caution.** Shares are unreliable when a neighbourhood has very few labelled homes: Vliegeniersbuurt has **one** labelled home, so it shows 100% A-label. The original top-5 could have been distorted by cases like this. From here on, a neighbourhood is analysed only if it has **at least 100 labelled homes and an income figure**. In practice the two rules exclude the same 8 areas, because Den Haag in Cijfers withholds income figures for areas with very few residents (Kerketuinen/Zichtenburg has 130 labelled homes but no income figure). Figure 3 shows where the cut falls.

````python
MIN_HOMES = 100

df["in_analysis"] = (df["labelled_homes"] >= MIN_HOMES) & df["income"].notna()

HIGH_INCOME = df.loc[df["in_analysis"], "income"].quantile(0.75)

df["high_income"] = df["income"] > HIGH_INCOME


# join to the boundaries and project to the Dutch national grid (RD New, metres) for mapping
nb = geo_raw.merge(df, left_on="neighbourhood", right_index=True, how="left").to_crs(28992)

an = df[df["in_analysis"]].copy()


print(f"In the analysis: {len(an)} of {len(df)} neighbourhoods "
      f"({an['labelled_homes'].sum() / df['labelled_homes'].sum():.1%} of labelled homes)")

print(f"High-income threshold (75th percentile): €{HIGH_INCOME:,.0f} → {an['high_income'].sum()} neighbourhoods")

print(f"New-build neighbourhoods in the analysis: {(an['area'] == NEW_BUILD).sum()}")
````

````python
homes = df["labelled_homes"].clip(lower=1)            # log scale: a count of 0 is drawn at 1

bins = np.logspace(0, np.log10(homes.max() * 1.1), 32)

left_out = df[~df["in_analysis"]].sort_values("labelled_homes")


fig, ax = plt.subplots(figsize=(13, 4.4))

ax.hist(homes[df["in_analysis"]], bins=bins, color="#b9b8b2", label=f"Kept ({df['in_analysis'].sum()})")

ax.hist(homes[~df["in_analysis"]], bins=bins, color=ORANGE, label=f"Left out ({len(left_out)})")

ax.axvline(MIN_HOMES, color=INK, ls="--", lw=1.2)

ax.text(MIN_HOMES * 1.08, ax.get_ylim()[1] * 0.92, f"{MIN_HOMES} labelled homes", fontsize=8.5, color=INK)


# list the excluded areas and the reason
lines = []

for name, r in left_out.iterrows():

    reason = "no income figure" if pd.isna(r["income"]) else "few homes"

    unit = "home" if r["labelled_homes"] == 1 else "homes"

    lines.append(f"{short_name(name)}: {r['labelled_homes']:,.0f} {unit}, {reason}")

ax.text(0.01, 0.97, "Left out\n" + "\n".join(lines), transform=ax.transAxes, va="top",
        fontsize=8, color=MUTED, linespacing=1.35)

ax.set_xscale("log")

ax.set_xlabel("Labelled homes per neighbourhood (log scale)")

ax.set_ylabel("Neighbourhoods")

ax.legend(loc="upper right")

ax.grid(axis="x", visible=False)

save_figure(fig, "small_counts", "8 areas have almost no labelled homes or no income figure, so they are left out", y=1.02)
````

**What is left out, and where?**

````python
coverage = nb.assign(labelled_homes=nb["labelled_homes"].where(nb["in_analysis"]))


fig, ax = plt.subplots(figsize=(8, 7.5))

choropleth(coverage, "labelled_homes", ax, "Greys", k=4, fmt="{:,.0f}", title="Homes with a label")

pts = nb[~nb["in_analysis"]].geometry.representative_point()

for x, y, name in zip(pts.x, pts.y, nb.loc[~nb["in_analysis"], "neighbourhood"]):
    ax.annotate(short_name(name), (x, y), fontsize=7, color=ORANGE, ha="center", fontweight="bold")

save_figure(fig, "coverage_map",
            f"{(~nb['in_analysis']).sum()} areas are left out: almost no homes or no income figure", y=0.97)
````

> **Insight.** The 8 excluded areas are almost empty (Figures 3 and 4): the Oostduinen dune reserve, a few industrial and edge zones in the new-build district, and the Kerketuinen business park. Together they hold only **0.1%** of labelled homes, so leaving them out does not change the picture for the city.

## 4. Where are the good labels?

**First, the city as a whole.** The share of each label across all labelled homes, and how many neighbourhoods have a majority at either end of the scale.

> **Why this step.** Before comparing neighbourhoods, it helps to know the baseline: what a "typical" home in The Hague looks like, and whether good or poor labels are the norm.

````python
city_mix = an[LABELS].sum() / an["labelled_homes"].sum() * 100

city_A = city_mix[["A++", "A+", "A"]].sum()

city_EG = city_mix[["E", "F", "G"]].sum()

majority = pd.DataFrame({
    "More than half\nA or better": an.groupby("area")["pct_A_or_better"].apply(lambda s: (s > 50).sum()),
    "More than half\nE, F or G": an.groupby("area")["pct_E_to_G"].apply(lambda s: (s > 50).sum()),
}).T.reindex(columns=[NEW_BUILD, REST], fill_value=0)


fig, (ax, ax2) = plt.subplots(1, 2, figsize=(14, 4.4), width_ratios=[1.2, 1])

# left: city-wide label mix
ax.bar(LABELS, city_mix, color=LABEL_COLORS, width=0.75)

for i, v in enumerate(city_mix):
    ax.text(i, v + 0.5, f"{v:.0f}%", ha="center", fontsize=8.5, color=MUTED)

ax.yaxis.set_major_formatter(pct)

ax.set_ylabel("Share of labelled homes")

ax.set_title(f"All {len(an)} neighbourhoods: {city_A:.0f}% A or better, {city_EG:.0f}% E–G")

ax.grid(axis="x", visible=False)

# right: neighbourhoods with a majority at either end, split by area
left = np.zeros(len(majority))

for area in [NEW_BUILD, REST]:

    ax2.barh(range(len(majority)), majority[area], left=left, color=AREA_COLORS[area], height=0.55, label=area)

    left += majority[area].to_numpy()

for i, total in enumerate(left):
    ax2.text(total + 0.8, i, f"{total:.0f} neighbourhoods", va="center", fontsize=9, color=INK)

ax2.set_yticks(range(len(majority)), majority.index)

ax2.invert_yaxis()

ax2.set_xlim(0, left.max() * 1.35)

ax2.set_xlabel("Neighbourhoods")

ax2.set_title("Where one end of the scale is the majority")

ax2.legend(loc="upper right", fontsize=8.5)

ax2.grid(axis="y", visible=False)

save_figure(fig, "city_label_mix", "Poor labels are the norm: G is the most common label in the city", y=1.03, tight=True)
````

> **Insight.** Across the 106 neighbourhoods, **15% of labelled homes are A or better and 45% are E, F or G**; G alone is the most common label (24%). Only **9** neighbourhoods have more than half their homes at A or better, and 8 of them are in the new-build district. By contrast, **50** neighbourhoods have more than half their homes at E–G, none of them new-build (Figure 5).

**Next, the full label mix per neighbourhood.** Each thin bar is one neighbourhood, split into its nine label classes and sorted from best average label (left) to worst (right).

> **Why this step.** Grouping nine classes into three hides a lot. This chart shows every class for every neighbourhood at once, so it is easy to see whether the city is split into a few very good areas and many poor ones, or whether labels change gradually.

````python
order = an.sort_values("label_score", ascending=False)

sh = order[LABELS].div(order["labelled_homes"], axis=0) * 100

x = np.arange(len(sh))


fig, (ax, ax2) = plt.subplots(2, 1, figsize=(13, 5.6), height_ratios=[4, 1], sharex=True)

# top: stacked label shares
bottom = np.zeros(len(sh))

for lab, col in zip(LABELS, LABEL_COLORS):

    ax.bar(x, sh[lab], bottom=bottom, color=col, width=1.0, label=lab, linewidth=0)

    bottom += sh[lab].to_numpy()

ax.set_ylim(0, 100)

ax.yaxis.set_major_formatter(pct)

ax.grid(False)

ax.set_ylabel("Share of labelled homes")

ax.legend(ncol=9, loc="upper left", bbox_to_anchor=(0, 1.13), fontsize=9, handlelength=1, columnspacing=1)

# bottom: income, new-build neighbourhoods in aqua
is_new = (order["area"] == NEW_BUILD).to_numpy()

ax2.bar(x, order["income"], color=np.where(is_new, AQUA, "#b9b8b2"), width=0.85)

ax2.yaxis.set_major_formatter(eur)

ax2.set_ylabel("Income")

ax2.set_xticks([])

ax2.set_xlabel("Neighbourhoods, best average label → worst     (aqua = new-build district)")

save_figure(fig, "label_mix_strip", "A few neighbourhoods are almost all A-label; most of the city is D or worse", y=1.04)
````

> **Insight.** The label mix changes gradually from left to right, but the far left is a small group of almost entirely A-label neighbourhoods, nearly all in the new-build district (aqua bars). The tallest income bars (bottom panel) are spread across the middle and right of the chart, not grouped on the left (Figure 6).

**Side by side: income and the two ends of the label scale.**

> **Why this step.** Putting the maps next to each other with the same quantile classes makes it easy to compare patterns by eye. If the hypothesis were right, the darkest income areas (left) would match the darkest green areas (middle) and the lightest red areas (right).

````python
view = nb.copy()

for c in ["income", "pct_A_or_better", "pct_E_to_G"]:
    view[c] = view[c].where(view["in_analysis"])


fig, axes = plt.subplots(1, 3, figsize=(17, 6.4))

choropleth(view, "income", axes[0], "Blues", fmt="€{:,.0f}", title="Income (quintiles)")

choropleth(view, "pct_A_or_better", axes[1], "Greens", fmt="{:.0f}%", title="% A-label or better")

choropleth(view, "pct_E_to_G", axes[2], "Reds", fmt="{:.0f}%", title="% E, F or G")

for ax, t in zip(axes, ["Household income", "Homes with label A or better", "Homes with label E–G"]):
    ax.set_title(t, fontsize=11)

save_figure(fig, "maps_side_by_side",
            "The richest areas, along the dunes in the north-west, have the most E–G homes", y=1.0, tight=True)
````

> **Insight.** The income map and the E–G map look alike (Figure 7): the dark-blue dune-side areas in the north-west (Benoordenhout, Statenkwartier, Archipelbuurt, Vogelwijk) are also dark red. The A-label map shows a different geography: the new-build district in the east, Hoge Veld and Kraayenstein in the south-west, and the redeveloped strip around Laakhaven and Binckhorst near the centre. This is the first sign that income and good labels are *not* found in the same places.

**One map, two variables: a bivariate choropleth.**

> **Why this step.** Three separate maps still make the reader compare by eye. A bivariate map puts both variables in one colour: blue for income, pink for good labels. If income and good labels went together, most areas would be either light grey (low–low) or dark purple (high–high). The legend shows how many neighbourhoods fall in each of the nine cells.

````python
BIV = np.array([["#e8e8e8", "#dfb0d6", "#be64ac"],     # rows: income tercile low → high
                ["#ace4e4", "#a5add3", "#8c62aa"],     # cols: % A-or-better tercile low → high
                ["#5ac8c8", "#5698b9", "#3b4994"]])

view = nb[nb["in_analysis"]].copy()

inc_t = pd.qcut(view["income"], 3, labels=False)

lab_t = pd.qcut(view["pct_A_or_better"], 3, labels=False)

view["biv"] = [BIV[i, j] for i, j in zip(inc_t, lab_t)]

counts = pd.crosstab(inc_t, lab_t)


fig, ax = plt.subplots(figsize=(9, 8))

nb[~nb["in_analysis"]].plot(ax=ax, color=NODATA, edgecolor="white", linewidth=0.4, hatch="////")

view.plot(ax=ax, color=view["biv"].tolist(), edgecolor="white", linewidth=0.4)

map_axis(ax)


# 3 × 3 legend with the number of neighbourhoods in each cell
lax = fig.add_axes([0.14, 0.63, 0.19, 0.19])

for i in range(3):

    for j in range(3):

        lax.add_patch(plt.Rectangle((j, i), 1, 1, color=BIV[i, j]))

        lax.text(j + 0.5, i + 0.5, counts.iat[i, j], ha="center", va="center", fontsize=9,
                 fontweight="bold", color="white" if i + j >= 3 else INK)

lax.set_xlim(0, 3)

lax.set_ylim(0, 3)

lax.set_aspect("equal")

lax.grid(False)

lax.set_xticks([])

lax.set_yticks([])

lax.set_xlabel("% A or better →", fontsize=8)

lax.set_ylabel("Income →", fontsize=8)

lax.set_title("Neighbourhoods per cell", fontsize=8, fontweight="normal", color=MUTED)

for s in lax.spines.values():
    s.set_visible(False)

save_figure(fig, "bivariate_map", "No diagonal: rich areas are as likely to have few A-labels as many", y=0.95)


print("Neighbourhoods per cell (rows: income tercile low → high, cols: % A-or-better tercile low → high)")

print(counts.to_string())
````

> **Insight.** There is no diagonal (Figure 8). If the hypothesis held, most neighbourhoods would sit in the light-grey (low–low) and dark-indigo (high–high) corners. Instead, each of the nine combinations holds 8 to 17 neighbourhoods. The north-west is mostly **teal** (rich, few A-labels), the inner south is mostly **pink** (lower income, more A-labels after renewal), and the dark-indigo areas are mainly in the new-build district, the south-west edge and a few park-side areas in the north.

## 5. Testing the hypothesis

**Income against the share of A-label homes.**

> **Why this step.** A scatter plot shows the whole relationship, not just a summary. The two groups are coloured separately because the maps already hinted that the new-build district behaves differently from the rest of the city. **Spearman's rank correlation** is used rather than Pearson because income is skewed (a few very rich areas) and label shares are bounded at 0 and 100%.

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5.6), sharex=True)

panels = [("pct_A_or_better", "Homes with label A or better"),
          ("pct_E_to_G", "Homes with label E, F or G")]

for ax, (col, ylab) in zip(axes, panels):

    for area in [REST, NEW_BUILD]:

        s = an[an["area"] == area]

        rho, p = spearmanr(s["income"], s[col])

        ax.scatter(s["income"], s[col], s=np.sqrt(s["labelled_homes"]) * 2.2, color=AREA_COLORS[area],
                   alpha=0.75, edgecolor="white", linewidth=0.6,
                   label=f"{area} (n = {len(s)}, ρ = {rho:+.2f})")

    rho, p = spearmanr(an["income"], an[col])

    ax.axvline(HIGH_INCOME, color=ORANGE, ls="--", lw=1.2)

    ax.text(HIGH_INCOME, 101, " high income →", color=ORANGE, fontsize=8, va="bottom")

    ax.set_ylim(-3, 108)

    ax.yaxis.set_major_formatter(pct)

    ax.xaxis.set_major_formatter(eur)

    ax.set_xlabel("Average standardised disposable household income")

    ax.set_title(f"{ylab}   ·   all neighbourhoods ρ = {rho:+.2f} (p = {fmt_p(p)})", fontsize=10)

    ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.13), fontsize=8, ncol=1)


# label the top-5 A-label areas and the two richest
for name in an.nlargest(5, "pct_A_or_better").index.tolist() + ["74 Duttendel", "68 Marlot"]:

    r = an.loc[name]

    axes[0].annotate(short_name(name), (r["income"], r["pct_A_or_better"]), xytext=(5, 2),
                     textcoords="offset points", fontsize=7, color=MUTED)

save_figure(fig, "scatter_income_labels",
            "Income does not predict good labels; outside the new-build district it predicts bad ones", y=1.02, tight=True)
````

> **Insight.** Citywide, the link between income and A-label share is weak and not significant (ρ = −0.14, p = 0.16), which is what the original notebook found. The colours in Figure 9 show why. The **new-build neighbourhoods** (aqua) sit high on the A-label axis at middle incomes. In the **rest of the city** (grey), the link is negative: richer areas have fewer A-label homes (ρ = −0.34) and many more E–G homes (ρ = +0.47). The richest neighbourhoods (Duttendel at €80,900, Marlot at €64,700) have almost no A-label homes.

**The label mix by income quartile.**

> **Why this step.** Quartiles turn a cloud of points into four groups that are easy to compare. Showing the city with and without the new-build district side by side tests whether the result depends on the 15 new-build neighbourhoods.

````python
def quartile_mix(frame):
    """Pooled label mix and income range per income quartile."""

    q = pd.qcut(frame["income"], 4, labels=["Q1 lowest", "Q2", "Q3", "Q4 highest"])

    totals = frame.groupby(q, observed=True)[LABELS].sum()          # pool homes within each quartile

    income_range = frame.groupby(q, observed=True)["income"].agg(["min", "max"])

    return totals.div(totals.sum(axis=1), axis=0) * 100, income_range


fig, axes = plt.subplots(1, 2, figsize=(14, 4.6), sharey=True)

frames = [("All neighbourhoods", an),
          ("Without the new-build district", an[an["area"] == REST])]

for ax, (title, frame) in zip(axes, frames):

    mix, income_range = quartile_mix(frame)

    stacked_label_bars(ax, mix, LABELS, LABEL_COLORS)

    best = mix[["A++", "A+", "A"]].sum(axis=1)

    worst = mix[["E", "F", "G"]].sum(axis=1)

    for i, (a, e) in enumerate(zip(best, worst)):

        ax.text(1, i, f"≥A: {a:.0f}%", va="center", fontsize=8, color="white", fontweight="bold")

        ax.text(99, i, f"E–G: {e:.0f}%", va="center", ha="right", fontsize=8, color="white", fontweight="bold")

    ticks = [f"{q}  €{lo / 1000:.0f}–{hi / 1000:.0f}k" for q, (lo, hi) in zip(mix.index, income_range.to_numpy())]

    ax.set_yticks(range(len(mix)), ticks)

    ax.set_title(f"{title} (n = {len(frame)})")

axes[0].invert_yaxis()

axes[1].legend(ncol=9, loc="upper right", bbox_to_anchor=(1, -0.1), fontsize=8, handlelength=1)

save_figure(fig, "quartile_label_mix",
            "The richest quarter has the worst label mix, with or without the new-build district", y=1.03, tight=True)
````

> **Insight.** In the full city, the **richest quarter has the worst mix** (Figure 10): 62% of its labelled homes are E–G, against 39% in the poorest quarter. The third quarter looks best only because most new-build neighbourhoods fall in the €28–36k range. With them removed (right panel), the trend is steady: the E–G share rises from 31% in the poorest quarter to 53%, 55% and 68% as income goes up, while the share of A-or-better falls from 15% to 6%.

**High-income neighbourhoods vs the rest: a direct test of H1.**

> **Why this step.** The hypothesis is about *high-income* neighbourhoods specifically, so this compares the top quarter with the other three quarters on all three measures. The **Mann-Whitney U test** asks whether values in one group tend to be larger than in the other, without assuming normal distributions. Each dot in Figure 11 is one neighbourhood; the black line is the median and the grey bar the middle half.

````python
H1_MEASURES = [("pct_A_or_better", "% A-label or better"),
               ("pct_E_to_G", "% E–G"),
               ("label_score", "Average label score (1–9)")]

rows = []

for col, label in H1_MEASURES:

    hi = an.loc[an["high_income"], col]

    lo = an.loc[~an["high_income"], col]

    u, p = mannwhitneyu(hi, lo, alternative="two-sided")

    higher_is_better = col != "pct_E_to_G"

    better = (hi.median() > lo.median()) == higher_is_better

    rows.append({"measure": label,
                 "high income (median)": hi.median(),
                 "rest (median)": lo.median(),
                 "direction": "high income better" if better else "high income worse",
                 "Mann-Whitney p": fmt_p(p)})

h1_table = pd.DataFrame(rows).set_index("measure")

print(f"High income: n = {an['high_income'].sum()} · rest: n = {(~an['high_income']).sum()}")

h1_table
````

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.8))

for ax, (col, label) in zip(axes, H1_MEASURES):

    hi = an.loc[an["high_income"], col]

    lo = an.loc[~an["high_income"], col]

    strip_with_box(ax, [hi, lo], [ORANGE, REST_GREY],
                   [f"High income\n(n = {len(hi)})", f"Rest\n(n = {len(lo)})"])

    if col == "label_score":
        ax.set_yticks(range(1, 10), [f"{s} ({lab})" for s, lab in zip(range(1, 10), LABELS[::-1])], fontsize=8)
    else:
        ax.yaxis.set_major_formatter(pct)

    ax.set_title(f"{label}\nmedian {hi.median():.1f} vs {lo.median():.1f} · "
                 f"p = {h1_table.loc[label, 'Mann-Whitney p']}", fontsize=10)

save_figure(fig, "h1_high_income_vs_rest",
            "High-income neighbourhoods are worse on all three label measures", y=1.04, tight=True)
````

> **Insight.** H1 is **rejected** (Figure 11). High-income neighbourhoods (above €35,575, n = 27) are worse on every measure: median 3.9% A-or-better vs 9.7% for the rest, 70.5% E–G vs 39.7%, and an average label score of 3.1 (between F and E) vs 4.0 (D). All three differences are significant (p = 0.022, 0.002 and 0.013).

<a id="age"></a>
## 6. The building-age effect

The dataset has no construction year, but the **new-build district** flag is a good stand-in: Leidschenveen-Ypenburg was built almost entirely after 1995, under recent building codes. If housing age drives labels, the label measures should differ sharply between the district and the rest of the city, and the income–label link should look very different inside and outside it.

> **Why this step.** This is the step the original analysis skipped. Its top-5 list was made up of new-build areas, and the conclusion ("no correlation") came from mixing two groups that behave in opposite ways. Figure 12 compares the two groups directly; Figure 13 then shows how the income correlations change when they are separated.

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.8))

for ax, (col, label) in zip(axes, H1_MEASURES):

    new = an.loc[an["area"] == NEW_BUILD, col]

    old = an.loc[an["area"] == REST, col]

    u, p = mannwhitneyu(new, old, alternative="two-sided")

    strip_with_box(ax, [new, old], [AQUA, REST_GREY],
                   [f"New-build district\n(n = {len(new)})", f"Rest of the city\n(n = {len(old)})"])

    if col == "label_score":
        ax.set_yticks(range(1, 10), [f"{s} ({lab})" for s, lab in zip(range(1, 10), LABELS[::-1])], fontsize=8)
    else:
        ax.yaxis.set_major_formatter(pct)

    p_text = fmt_p(p) if p < 0.001 else "= " + fmt_p(p)

    ax.set_title(f"{label}\nmedian {new.median():.1f} vs {old.median():.1f} · p {p_text}", fontsize=10)

save_figure(fig, "new_build_vs_rest",
            "The new-build district has far better labels than the rest of the city", y=1.04, tight=True)
````

> **Insight.** The gap between the two groups is much larger than the gap between income groups (Figure 12 vs Figure 11). The median new-build neighbourhood has **51%** of its homes at A or better, against **6%** in the rest of the city, and almost no E–G homes. Housing age, not income, separates the good labels from the bad.

**Income correlations, inside and outside the new-build district.**

> **Why this step.** A **forest plot** puts every correlation with its 95% bootstrap confidence interval on one axis, so the size, direction and uncertainty of each one can be read at a glance.

````python
groups = {"All neighbourhoods": an,
          "Rest of the city": an[an["area"] == REST],
          "New-build district": an[an["area"] == NEW_BUILD]}

measures = {"pct_A_or_better": "% A or better",
            "label_score": "Average label score",
            "pct_E_to_G": "% E–G"}

res = []

for g, frame in groups.items():

    for col, mlabel in measures.items():

        rho, p, lo, hi = spearman_ci(frame["income"], frame[col])

        res.append({"group": g, "measure": mlabel, "n": len(frame),
                    "rho": rho, "ci_low": lo, "ci_high": hi, "p": fmt_p(p)})

corr = pd.DataFrame(res)


colors = {"All neighbourhoods": INK, "Rest of the city": REST_GREY, "New-build district": AQUA}

fig, ax = plt.subplots(figsize=(10, 5.2))

ypos = 0

yticks = []

for m in measures.values():

    for g in groups:

        r = corr[(corr["group"] == g) & (corr["measure"] == m)].iloc[0]

        ax.plot([r["ci_low"], r["ci_high"]], [ypos, ypos], color=colors[g], lw=2.5, solid_capstyle="round")

        ax.scatter(r["rho"], ypos, color=colors[g], s=60, zorder=3, edgecolor="white")

        ax.text(1.04, ypos, f"{r['rho']:+.2f}  (n = {r['n']})", va="center", fontsize=8, color=colors[g])

        yticks.append((ypos, f"{m} · {g}"))

        ypos -= 1

    ypos -= 0.6

ax.axvline(0, color=MUTED, lw=1)

ax.set_yticks([y for y, _ in yticks], [t for _, t in yticks], fontsize=8.5)

ax.set_xlim(-1, 1.0)

ax.set_xlabel("Spearman ρ with household income (95% bootstrap CI)")

ax.grid(axis="y", visible=False)

save_figure(fig, "forest_correlations",
            "In the existing city, richer means worse labels; the citywide 'no effect' is two effects cancelling", y=1.0)


corr.round(2).set_index(["measure", "group"])
````

> **Insight.** Outside the new-build district, all three correlations point the same way and none of their confidence intervals cross zero (Figure 13): higher income goes with fewer A-labels (ρ = −0.34), a lower average label (ρ = −0.41) and more E–G homes (ρ = +0.47). Inside the new-build district (n = 15), the intervals are wide and include zero: everyone there lives in recent housing, so income makes no clear difference. Combining the two groups gives the weak citywide result.

> **Caution.** The new-build flag is only a rough stand-in for building age. Other recently built or renovated areas (Hoge Veld, Kraayenstein, Laakhaven, Binckhorst) are counted in the 'rest of the city' group, which if anything *weakens* the negative link found there. A proper test would use the average construction year per neighbourhood, which Den Haag in Cijfers also publishes. Part of the label pattern may also reflect who registered labels: most older homes received a *provisional* label in 2015 based mainly on building type and year, while housing associations registered full labels after renovating their stock, which is concentrated in lower-income areas.

<a id="moran"></a>
## 7. Spatial clustering: do similar neighbourhoods sit together?

**Moran's I** measures whether neighbouring areas have similar values. It is about 0 when values are scattered randomly across the map, and positive when high values sit next to high values and low next to low. Two neighbourhoods count as neighbours when their boundaries touch (*queen contiguity*).

> **Why this step.** The maps suggest clear clusters: wealthy dune-side areas in the north-west and A-label new builds in the south-east. Moran's I tests this formally. It also matters for the correlations above: neighbouring areas are not fully independent observations, so p-values from ordinary tests are somewhat too optimistic when there is strong clustering. To get a p-value, the values are shuffled randomly over the map 999 times and the real I is compared with those shuffled ones.

````python
geo_an = geo_raw.merge(an, left_on="neighbourhood", right_index=True, how="inner").reset_index(drop=True)

W = queen_weights(geo_an)

n_neighbours = (W > 0).sum(axis=1)

print(f"Neighbours per area: mean {n_neighbours.mean():.1f}, min {n_neighbours.min()}, "
      f"max {n_neighbours.max()} · areas with none: {(n_neighbours == 0).sum()}")


moran = {}

fig, axes = plt.subplots(1, 3, figsize=(15, 3.8), sharey=True)

panels = [("income", "Household income", BLUE),
          ("pct_A_or_better", "% A-label or better", AQUA),
          ("pct_E_to_G", "% E–G", ORANGE)]

for ax, (col, label, color) in zip(axes, panels):

    I, p, sims = morans_i(geo_an[col], W)

    moran[label] = {"Moran's I": I, "p (999 permutations)": fmt_p(p), "expected if random": -1 / (len(W) - 1)}

    ax.hist(sims, bins=30, color="#c9c8c2")

    ax.axvline(I, color=color, lw=3)

    ax.text(I, ax.get_ylim()[1] * 0.9, f"  I = {I:.2f}", color=color, fontweight="bold", fontsize=10)

    ax.set_title(label)

    ax.set_xlabel("Moran's I")

    ax.set_xlim(-0.35, max(0.8, I + 0.1))

axes[0].set_ylabel("Shuffled maps")

save_figure(fig, "morans_i", "All three measures are strongly clustered on the map", y=1.06, tight=True)


pd.DataFrame(moran).T
````

> **Insight.** All three measures are strongly clustered (Figure 14): Moran's I is 0.58 for income, 0.59 for the A-label share and 0.57 for the E–G share. None of the 999 random shuffles came close (p = 0.001, the smallest possible with 999 shuffles). Neighbourhoods look like their neighbours, so the 106 areas are not fully independent observations. The p-values in sections 5 and 6 are therefore somewhat optimistic, although the main contrast (ρ = +0.47 for E–G outside the new-build district) is large enough that the conclusion is unlikely to change.

## Part II · Neighbourhood types by energy-label mix (K-means)

Part I compared neighbourhoods one measure at a time. Part II lets the data group them: **K-means clustering** sorts the neighbourhoods into a small number of *types* that have similar label mixes. The questions are:

1. What kinds of neighbourhoods does The Hague have, in terms of their label mix?
2. Where are they on the map?
3. Are the types any good: well separated, stable, and spatially coherent?
4. Do the types differ in household income? This links back to the hypothesis from Part I.

<a id="kchoice"></a>
## 9. Features and the number of clusters

**Features.** The share of labelled homes in each of seven classes: **A-or-better** (A++, A+ and A combined), B, C, D, E, F and G. Every feature is a percentage on the same 0–100 scale, so no further scaling is needed, and the cluster centres can be read directly as "the typical label mix of this type".

**Which neighbourhoods.** All 107 with at least 100 labelled homes. Income is *not* used to build the clusters, so neighbourhoods without an income figure can stay in. Income is only brought back in section 12 to describe the types.

> **Why this step.** K-means needs the number of clusters *k* in advance. Three checks are combined instead of relying only on the elbow plot: **inertia** (within-cluster spread, which always falls as k grows; look for the bend), the **silhouette score** (how much closer each neighbourhood is to its own cluster than to the next one; higher is better), and **stability** (do different random starts give the same clusters? measured with the adjusted Rand index, where 1 means identical). A good k is one where the elbow bends, the silhouette is high and the result does not depend on the random seed.

````python
cl = df[df["labelled_homes"] >= MIN_HOMES].copy()

X = cl[LABELS].div(cl["labelled_homes"], axis=0) * 100

X["≥A"] = X[["A++", "A+", "A"]].sum(axis=1)

X = X[FEATURES]

print(f"{len(X)} neighbourhoods × {X.shape[1]} features")


K_RANGE = range(2, 11)

scan = []

for k in K_RANGE:

    base = KMeans(n_clusters=k, n_init=50, random_state=42).fit(X)

    ari = [adjusted_rand_score(base.labels_, KMeans(n_clusters=k, n_init=50, random_state=s).fit(X).labels_)
           for s in range(1, 11)]

    scan.append({"k": k,
                 "inertia": base.inertia_,
                 "silhouette": silhouette_score(X, base.labels_),
                 "stability (min ARI)": min(ari),
                 "smallest cluster": np.bincount(base.labels_).min()})

scan = pd.DataFrame(scan).set_index("k")


fig, axes = plt.subplots(1, 3, figsize=(15, 3.8))

panels = [(axes[0], "inertia", MUTED, "Inertia (elbow)"),
          (axes[1], "silhouette", BLUE, "Silhouette score ↑"),
          (axes[2], "stability (min ARI)", AQUA, "Stability across 10 seeds ↑")]

for ax, col, color, title in panels:

    ax.plot(scan.index, scan[col], marker="o", color=color)

    ax.axvline(5, color=ORANGE, ls="--", lw=1.2)

    ax.set_title(title)

    ax.set_xlabel("Number of clusters k")

    ax.set_xticks(list(K_RANGE))

axes[0].text(5.15, scan["inertia"].max() * 0.95, "k = 5", color=ORANGE, fontsize=9)

axes[2].set_ylim(0.5, 1.05)

save_figure(fig, "choose_k",
            "k = 5 is stable and close to the best silhouette; k = 6 and k ≥ 8 depend on the random start", y=1.06, tight=True)


scan.round(3)
````

> **Insight.** The elbow plot has no sharp bend (Figure 16): inertia falls steadily, which is why the original notebook leaned towards 9 clusters. The other two checks are clearer. At **k = 5** the silhouette is 0.37, close to the best value, and all 10 random starts give exactly the same clusters (ARI = 1.00). k = 7 scores slightly higher (0.39), but it gets there by splitting the mixed and E-heavy groups into sub-types that differ by only a few percentage points. At k = 6 and at k = 8–10 the result changes with the random start (ARI 0.74–0.85), so those extra clusters are not robust. k = 5, the number the original notebook started with, is kept.

<a id="types"></a>
## 10. The five neighbourhood types

> **Why this step.** Cluster numbers from K-means are arbitrary (cluster 0 is not 'better' than cluster 3). To make the result readable, the clusters are **ordered by their average label score** and **named after their label mix**, and each type gets the label colour that matches its typical label. The names describe the label mix only, not the age or type of the housing.

````python
K = 5

km = KMeans(n_clusters=K, n_init=50, random_state=42).fit(X)

cl["cluster_raw"] = km.labels_


# order clusters from best to worst average label
score_by_cluster = cl.groupby("cluster_raw")["label_score"].mean().sort_values(ascending=False)

centres = pd.DataFrame(km.cluster_centers_, columns=FEATURES).loc[score_by_cluster.index]


def name_type(c):
    """Readable name from a cluster centre (shares in %)."""

    if c["≥A"] >= 60:
        return "Mostly A"

    if c["≥A"] + c["B"] >= 60:
        return "Mostly A and B"

    if c["G"] >= 35:
        return "G-dominated"

    if c["E"] >= 35:
        return "E-dominated"

    return "Mixed, centred on C"


TYPE_NAMES = [name_type(centres.loc[i]) for i in score_by_cluster.index]

assert len(set(TYPE_NAMES)) == K, TYPE_NAMES        # every cluster gets its own name

cl["type"] = pd.Categorical(cl["cluster_raw"].map(dict(zip(score_by_cluster.index, TYPE_NAMES))),
                            categories=TYPE_NAMES, ordered=True)

centres.index = TYPE_NAMES

TYPE_COLORS = dict(zip(TYPE_NAMES, [LABEL_COLORS[0], LABEL_COLORS[3], LABEL_COLORS[4],
                                    LABEL_COLORS[6], LABEL_COLORS[8]]))

sizes = cl["type"].value_counts().reindex(TYPE_NAMES)


# profile chart: the cluster centre = the typical label mix of each type
fig, ax = plt.subplots(figsize=(12, 3.8))

stacked_label_bars(ax, centres, FEATURES, FEATURE_COLORS, height=0.68, min_text=7, text_dark=["B", "C", "D"])

ax.set_yticks(range(K), [f"{t}  (n = {n})" for t, n in zip(TYPE_NAMES, sizes)])

ax.invert_yaxis()

ax.legend(ncol=7, loc="upper left", bbox_to_anchor=(0, 1.16), fontsize=9, handlelength=1)

save_figure(fig, "cluster_profiles",
            "Five types: two A/B types (mostly new-build), one mixed type, and two dominated by one poor label", y=1.08)


centres.round(1)
````

> **Insight.** The five types, from best to worst average label (Figure 17):
>
> * **Mostly A** (6 neighbourhoods): 85% A-or-better. Recent developments: four in Leidschenveen-Ypenburg, plus Hoge Veld and Binckhorst.
> * **Mostly A and B** (11): about 40% A and 50% B. The rest of the new-build district, plus Kraayenstein & Vroondaal.
> * **Mixed, centred on C** (36): no single label above 27%, with something in every class. The renewed inner-city and southern areas fall here.
> * **E-dominated** (18): almost half the homes are E.
> * **G-dominated** (36): 45% G and 17% F, with less than 11% of homes at B or better.
>
> The two poorest types cover 54 of the 107 neighbourhoods. The original notebook dropped A++ and A+ as near-empty; here they are merged into ≥A, so no homes are lost.

````python
geo_cl = nb.merge(cl[["type"]], left_on="neighbourhood", right_index=True, how="left")

known = geo_cl[geo_cl["type"].notna()]


fig, ax = plt.subplots(figsize=(9.5, 8.5))

geo_cl[geo_cl["type"].isna()].plot(ax=ax, color=NODATA, edgecolor="white", linewidth=0.4, hatch="////")

known.plot(ax=ax, color=[TYPE_COLORS[t] for t in known["type"]], edgecolor="white", linewidth=0.5)

map_axis(ax)

handles = [Patch(color=TYPE_COLORS[t], label=f"{t} ({sizes[t]})") for t in TYPE_NAMES]

handles.append(Patch(facecolor=NODATA, hatch="////", edgecolor="#9a9993", label="fewer than 100 labelled homes"))

ax.legend(handles=handles, loc="lower left", bbox_to_anchor=(0.98, 0.02), fontsize=9,
          title="Neighbourhood type", title_fontsize=9)

save_figure(fig, "cluster_map",
            "The types form zones: G in the pre-war centre and north-west, A/B on the new-build edges", y=0.95)
````

> **Insight.** Although K-means never saw the map, the types come out as clear zones (Figure 18). **G-dominated** neighbourhoods form one large block across the pre-war centre and the north-west (Statenkwartier, Archipelbuurt, Zeeheldenkwartier, the Bomen-/Bloemenbuurt, Benoordenhout) and reach down into Laakkwartier and Transvaal-Zuid. **Mixed** areas form a belt through the inner south (Schilderswijk, Transvaal, Moerwijk), where much housing was rebuilt during urban renewal. **E-dominated** neighbourhoods appear in patches, mostly in the post-war south-west (Bohemen, Leyenburg, Morgenstond) and the north-east (Duttendel, Uilennest, Kampen, Burgen en Horsten). The A and A/B types sit on the edges: the new-build east and the south-west tip.

<a id="evaluate"></a>
## 11. How good are the clusters?

Clustering has no "right answer" to score against, so the evaluation uses three angles:

| Check | Question |
|---|---|
| **Silhouette per neighbourhood** | Which neighbourhoods fit their type well, and which sit on the border between two types? |
| **Dominant label** | Do the types agree with the simple rule "colour each neighbourhood by its most common label", which the original notebook used? |
| **Spatial coherence** | Are neighbouring areas more often of the same type than chance would give? K-means never saw the map, so a yes here is independent evidence that the types are real. |

> **Why this step.** A single average silhouette can hide a few badly placed neighbourhoods. Looking at each one shows *where* the model is unsure, which is the clustering version of an error analysis.

````python
cl["silhouette"] = silhouette_samples(X, km.labels_)

cl["dominant_label"] = pd.Categorical(X.idxmax(axis=1), categories=FEATURES, ordered=True)


fig, (ax, ax2) = plt.subplots(1, 2, figsize=(15, 5.4), width_ratios=[1.35, 1])

# left: silhouette of every neighbourhood, grouped by type
y0 = 0

for t in TYPE_NAMES:

    s = cl.loc[cl["type"] == t, "silhouette"].sort_values()

    ax.barh(np.arange(y0, y0 + len(s)), s, color=TYPE_COLORS[t], height=1.0)

    ax.text(-0.02, y0 + len(s) / 2, t, ha="right", va="center", fontsize=9, color=INK)

    y0 += len(s) + 3

avg = cl["silhouette"].mean()

ax.axvline(avg, color=INK, ls="--", lw=1)

ax.text(avg, y0, f" mean {avg:.2f}", fontsize=9, va="bottom")

ax.axvline(0, color=MUTED, lw=0.8)

ax.set_yticks([])

ax.grid(axis="y", visible=False)

ax.set_xlabel("Silhouette (−1 wrong cluster · 0 on the border · 1 clear fit)")

ax.set_title("Fit of each neighbourhood to its type")

# right: type vs most common label
ct = pd.crosstab(cl["type"], cl["dominant_label"]).reindex(columns=FEATURES, fill_value=0)

ax2.imshow(ct.to_numpy(), cmap="Greys", aspect="auto", vmin=0, vmax=ct.to_numpy().max() * 1.6)

for i in range(ct.shape[0]):

    for j in range(ct.shape[1]):

        v = ct.iat[i, j]

        if v:
            ax2.text(j, i, v, ha="center", va="center", fontsize=10, fontweight="bold", color=INK)

ax2.set_xticks(range(len(FEATURES)), FEATURES)

ax2.set_yticks(range(K), TYPE_NAMES)

ax2.set_xlabel("Most common label in the neighbourhood")

ax2.grid(False)

ax2.set_title("Type vs most common label")

save_figure(fig, "cluster_evaluation", "Most neighbourhoods fit their type; the border cases are mixed areas", y=1.02, tight=True)


print("Five weakest fits (lowest silhouette):")

cl.nsmallest(5, "silhouette")[["type", "dominant_label", "silhouette"]].join(X.round(0))
````

> **Insight.** The average silhouette is 0.37, and 103 of the 107 neighbourhoods are positive, so they are closer to their own type than to any other (Figure 19, left). The four weakest fits are all in the **mixed** type. Belgisch Park, Noordpolderbuurt and Rond de Energiecentrale have slightly negative scores because they combine a lot of C *and* 30–36% G, which puts them between the mixed and G-dominated types. These are the model's 'errors'.
>
> The right panel compares the types with the original rule of colouring by the most common label. The two agree for the clear-cut types: 33 of the 36 G-dominated neighbourhoods have G as their most common label (the other 3 have F), and 17 of the 18 E-dominated have E. The mixed type is what K-means adds: 36 neighbourhoods whose most common label ranges from A to G, which the most-common-label rule would have scattered across six different colours.

**Spatial coherence.** The share of neighbouring pairs that are the same type, compared with 999 random shuffles of the types over the map.

````python
geo_k = geo_raw.merge(cl[["type"]], left_on="neighbourhood", right_index=True, how="inner").reset_index(drop=True)

A = (queen_weights(geo_k) > 0).astype(int)

types = geo_k["type"].astype(str).to_numpy()

pairs = np.argwhere(np.triu(A))


def same_share(t):
    """Share of neighbouring pairs whose two areas have the same type."""

    return (t[pairs[:, 0]] == t[pairs[:, 1]]).mean()


observed = same_share(types)

rng = np.random.default_rng(42)

shuffled = np.array([same_share(rng.permutation(types)) for _ in range(999)])

p_same = (1 + (shuffled >= observed).sum()) / 1000


print(f"Neighbouring pairs: {len(pairs)}")

print(f"Same type: {observed:.0%} of pairs  ·  random shuffles: {shuffled.mean():.0%} on average "
      f"(max {shuffled.max():.0%})  ·  p = {fmt_p(p_same)}")
````

````python
fig, ax = plt.subplots(figsize=(11, 3.8))

ax.hist(shuffled * 100, bins=30, color="#c9c8c2", label="999 random shuffles of the types")

ax.axvline(shuffled.mean() * 100, color=MUTED, ls="--", lw=1.2)

ax.text(shuffled.mean() * 100, ax.get_ylim()[1] * 0.92, f"  chance: {shuffled.mean():.0%}", color=MUTED, fontsize=9)

ax.axvline(observed * 100, color=BLUE, lw=3)

ax.text(observed * 100, ax.get_ylim()[1] * 0.92, f"  observed: {observed:.0%}", color=BLUE,
        fontweight="bold", fontsize=10)

ax.xaxis.set_major_formatter(pct)

ax.set_xlim(10, max(60, observed * 100 + 8))

ax.set_xlabel(f"Neighbouring pairs of the same type (of {len(pairs)} pairs)")

ax.set_ylabel("Shuffled maps")

ax.legend(loc="upper left", bbox_to_anchor=(0, 0.8), fontsize=8.5)

save_figure(fig, "spatial_coherence",
            "Neighbours share a type twice as often as chance, although K-means never saw the map", y=1.03)
````

> **Insight.** Half (50%) of all neighbouring pairs are of the same type (Figure 20). When the types are shuffled randomly over the map, the average is 26% and the highest of 999 shuffles is 37% (p = 0.001). The clustering used only label shares, so this spatial pattern is independent evidence that the types reflect real differences in the housing stock of different parts of the city.

<a id="typeincome"></a>
## 12. Do the types differ in income?

> **Why this step.** This links Part II back to the hypothesis. If richer neighbourhoods had better labels, the A and A/B types would have the highest incomes and the G type the lowest. The **Kruskal-Wallis test** checks whether income differs across the five types at all, without assuming normal distributions. Each dot is a neighbourhood, so the spread inside each type stays visible.

````python
inc = cl.dropna(subset=["income"])

groups_inc = [inc.loc[inc["type"] == t, "income"] for t in TYPE_NAMES]

H, p_kw = kruskal(*groups_inc)


fig, ax = plt.subplots(figsize=(12, 4.6))

rng = np.random.default_rng(1)

for i, (t, g) in enumerate(zip(TYPE_NAMES, groups_inc)):

    ax.scatter(g, i + rng.uniform(-0.18, 0.18, len(g)), s=38, color=TYPE_COLORS[t], alpha=0.8,
               edgecolor="white", linewidth=0.5)

    q1, med, q3 = g.quantile([0.25, 0.5, 0.75])

    ax.plot([q1, q3], [i, i], color=INK, lw=6, alpha=0.18, solid_capstyle="round")

    ax.scatter(med, i, marker="|", s=500, color=INK, linewidth=2.5, zorder=3)

    ax.text(g.max() + 1500, i, f"median €{med:,.0f}  (n = {len(g)})", va="center", fontsize=8.5, color=MUTED)

ax.axvline(HIGH_INCOME, color=ORANGE, ls="--", lw=1.2)

ax.text(HIGH_INCOME, K - 0.45, " high-income threshold", color=ORANGE, fontsize=8)

ax.set_yticks(range(K), TYPE_NAMES)

ax.invert_yaxis()

ax.xaxis.set_major_formatter(eur)

ax.set_xlabel("Average standardised disposable household income")

ax.set_xlim(None, inc["income"].max() + 20000)

p_text = fmt_p(p_kw) if p_kw < 0.001 else "= " + fmt_p(p_kw)

ax.set_title(f"Kruskal-Wallis H = {H:.1f}, p {p_text}", fontsize=10)

save_figure(fig, "income_by_type",
            "The A types have middle incomes; the richest areas fall in the G- and E-dominated types", y=1.04)
````

````python
summary_types = (inc.groupby("type", observed=True)
                    .agg(neighbourhoods=("income", "size"),
                         median_income=("income", "median"),
                         high_income=("high_income", "sum"),
                         new_build=("area", lambda s: (s == NEW_BUILD).sum())))


fig, axes = plt.subplots(1, 2, figsize=(14, 3.8), sharey=True)

panels = [("high_income", f"High-income neighbourhoods (of {int(summary_types['high_income'].sum())})"),
          ("new_build", f"New-build neighbourhoods (of {int(summary_types['new_build'].sum())})")]

for ax, (col, title) in zip(axes, panels):

    values = summary_types[col]

    ax.barh(range(K), values, color=[TYPE_COLORS[t] for t in TYPE_NAMES], height=0.6)

    for i, (v, n) in enumerate(zip(values, summary_types["neighbourhoods"])):
        ax.text(v + 0.3, i, f"{v:.0f} of {n} in this type", va="center", fontsize=8.5, color=MUTED)

    ax.set_xlim(0, values.max() * 1.6)

    ax.set_xlabel("Neighbourhoods")

    ax.set_title(title)

    ax.grid(axis="y", visible=False)

axes[0].set_yticks(range(K), TYPE_NAMES)

axes[0].invert_yaxis()

save_figure(fig, "type_composition",
            "Nearly half the high-income neighbourhoods are G-dominated; the A types are the new-build district", y=1.05, tight=True)


summary_types
````

> **Insight.** Income does differ between the types (Kruskal-Wallis H = 19.4, p < 0.001), but not in the direction the hypothesis expects (Figure 21). The two A types have **middle** incomes (medians €32,850 and €32,600). The **lowest** median is in the mixed type (€24,550), which holds the renewed areas of Schilderswijk, Transvaal and Moerwijk. The **G-dominated** type holds 13 of the 27 high-income neighbourhoods (Figure 22). It also has the widest income range (€20,500–68,700), because pre-war housing is home both to wealthy villa areas and to low-income areas such as Laakkwartier. The richest neighbourhood of all, Duttendel (€80,900), is E-dominated. This confirms Part I: labels follow the housing stock, and income does not predict which type a neighbourhood belongs to.

## 13. Findings and limitations

> **Answer to the research question.** High-income neighbourhoods in The Hague do **not** have more energy-efficient homes. Across the city there is no clear link (ρ = −0.14 for the share of A-label homes, Figure 9). Outside the post-1995 new-build district, richer neighbourhoods have clearly **worse** labels (ρ = +0.47 with the E–G share, Figure 13). The high-income quarter has a median of 71% E–G homes, against 40% for the rest (Figure 11).

**What drives labels instead.** The best labels are where the housing is newest: Leidschenveen-Ypenburg, plus new or redeveloped areas like Hoge Veld and Binckhorst (Figures 12 and 15). The worst labels are in the pre-war villa and townhouse areas near the dunes, which are also where incomes are highest (Figure 7). Income and labels both cluster strongly on the map (Moran's I ≈ 0.6, Figure 14), so this is a story about **different parts of the city**, not about individual households.

**Policy angle.** The usual worry is that poor households live in poorly insulated homes. In The Hague the largest stock of E–G homes is in wealthy neighbourhoods, where owners can pay for insulation themselves. The five retrofit candidates in section 8 have 86–94% E–G homes and incomes of €46,000–81,000 (Figure 15).

**Neighbourhood types (Part II).** K-means with k = 5 gives five stable, interpretable types (silhouette 0.37; identical across 10 random starts, Figure 16). The G-dominated type (36 neighbourhoods) is where retrofitting matters most, and it includes most of the city's wealthiest areas (Figures 21 and 22). The mixed type (36) holds the lowest-income areas and is the least clear-cut (all four negative silhouettes are there, Figure 19). The types are spatially coherent even though the model never saw the map (Figures 18 and 20).

> **Limitations**
>
> * **K-means on shares.** Label shares add up to 100%, so the seven features are not independent, and K-means treats a 10-point gap in A the same as a 10-point gap in G. A log-ratio transform or a model-based method (Gaussian mixture) would be worth comparing. The clusters weigh each neighbourhood equally, whatever its number of homes.
> * **Unknown reference year.** The export does not state which year the labels and incomes refer to. The original assignment was done in 2020, so the data is probably from 2018–2019.
> * **Registered labels only.** Shares are of homes *with* a label. Older owner-occupied homes often carry only a provisional label based on building type and year, and that is part of why labels follow building age so closely.
> * **Ecological data.** These are neighbourhood averages. They say nothing about whether an individual rich household lives in a better or worse home (the *ecological fallacy*).
> * **No building-age variable.** The new-build flag is a stand-in. Adding the average construction year per neighbourhood would allow a proper partial correlation.
> * **Spatial dependence.** Neighbourhoods are clustered (section 7), so the ordinary p-values are somewhat too small. A spatial regression would handle this properly.
